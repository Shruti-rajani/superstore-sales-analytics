```python
import pandas as pd
import numpy as np

# Load messy data fresh
df = pd.read_csv('messy_sales_data.csv')
print("Starting shape:", df.shape)

# Step 1 - Remove duplicates
df = df.drop_duplicates()
print("After duplicates removed:", df.shape)

# Step 2 - Drop missing Customer Names
df = df.dropna(subset=['Customer Name'])
print("After dropping missing names:", df.shape)

# Step 3 - Fill missing Region
df['Region'] = df['Region'].fillna(df['Region'].mode()[0])

# Step 4 - Fill missing Sales
df['Sales'] = df['Sales'].fillna(df['Sales'].median())

# Step 5 - Cap outliers
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + 1.5 * IQR
df['Sales'] = df['Sales'].clip(upper=upper_limit)

# Final check
print("Final shape:", df.shape)
print("Duplicates:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())

# Save properly
df.to_csv('clean_sales_data.csv', index=False)
print("Saved successfully.")
```
