```python
import pandas as pd
import numpy as np

df = pd.read_csv('messy_sales_data.csv')
print("Messy dataset shape:", df.shape)
df.head()
```

```python
df = df.drop_duplicates()
print("After removing duplicates: ", df.shape)
```

```python
df = df.dropna(subset=['Customer Name'])
print("After dropping missing customer name:", df.shape)
```

```python
region_mode = df['Region'].mode()[0]
df["Region"] =  df['Region'].fillna(region_mode)
print("Missing Region values remaining:", df['Region'].isnull().sum())
print(f"Filled with:{region_mode}")
```

```python
sales_median = df['Sales'].median()
df['Sales'] = df['Sales'].fillna(sales_median)
print("Missing Sales values remaining:", df['Sales'].isnull().sum())
print(f"Filled with median: {sales_median:.2f}")
```

```python
Q1 = df["Sales"].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3-Q1

upper_limit = Q3 + 1.5 * IQR

#Anything above Q3 + 1.5×IQR is considered an outlier.
print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Upper limit: {upper_limit:.2f}")
print(f"Outliers found: {(df['Sales'] > upper_limit).sum()}")

# Cap outliers at upper limit instead of dropping
df['Sales'] = df['Sales'].clip(upper=upper_limit)
print("Outliers handled.")
```

```python
print("=" * 40)
print("CLEANING SUMMARY")
print("=" * 40)
print(f"Final shape: {df.shape}")
print(f"Missing values remaining:\n{df.isnull().sum()[df.isnull().sum() > 0]}")
print(f"Duplicates remaining: {df.duplicated().sum()}")
```

```python
df.to_csv('clean_sales_data.csv', index=False)
print("Clean dataset saved.")
```
